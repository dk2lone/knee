```python
import time
from pathlib import Path

import pandas as pd

t0 = time.time()

LABELS = [
    "ACL", "MCL", "Medial Meniscus", "Lateral Meniscus",
    "Medial OA", "Lateral OA", "PF OA", "Effusion",
    "Synovitis", "Baker's", "Contusion", "Fracture",
]


def find_root():
    """Find the competition mount.

    The path is not always the same depth, so search for the directory that holds
    both test.csv and test_series/ rather than hardcoding one.
    """
    for c in [Path("/kaggle/input/rsna-knee-abnormality-detection"),
              Path("/kaggle/input/competitions/rsna-knee-abnormality-detection")]:
        if (c / "test.csv").is_file():
            return c
    base = Path("/kaggle/input")
    for d in sorted(p for p in base.iterdir() if p.is_dir()):
        if (d / "test.csv").is_file():
            return d
    raise FileNotFoundError("competition mount not found")


root = find_root()
test = pd.read_csv(root / "test.csv")
print(f"test studies: {len(test)}")

sub = test[["StudyInstanceUID"]].copy()
for c in LABELS:
    sub[c] = 0.5
sub.to_csv("submission.csv", index=False)

print(sub.shape)
print(sub.head())
print(f"runtime: {time.time() - t0:.1f}s")
```
